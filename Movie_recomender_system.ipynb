```python
#changing dict to list
import ast
def convert(obj):
  list=[]

  for i in ast.literal_eval(obj):
        list.append(i['name'])
  return list
```

```python
moviesandcredit_df['genres']=moviesandcredit_df['genres'].apply(convert)
```

```python
moviesandcredit_df['keywords']=moviesandcredit_df['keywords'].apply(convert)
```

```python
import ast
def convertandstopatat_3(obj):
  list=[]
  counter=0

  for i in ast.literal_eval(obj):
    if counter!=3:
      list.append(i['name'])
      counter+=1
    else:
      break
  return list
```

```python
moviesandcredit_df.iloc[0].cast
```

```python
moviesandcredit_df['cast']=moviesandcredit_df['cast'].apply(convertandstopatat_3)
```

```python
import ast
def fetch_director(obj):
  list=[]
  for i in ast.literal_eval(obj):
    if i["job"]=="Director":
      list.append(i["name"])
  else:
    pass
  return list
```

```python
moviesandcredit_df['crew']=moviesandcredit_df['crew'].apply(fetch_director)
```

```python

moviesandcredit_df['overview']=moviesandcredit_df['overview'].apply(lambda x:x.split())
```

```python
moviesandcredit_df.head()
```

```python
moviesandcredit_df['genres']=moviesandcredit_df['genres'].apply(lambda x:[i.replace(" ","") for i in x])
moviesandcredit_df['keywords']=moviesandcredit_df['keywords'].apply(lambda x:[i.replace(" ","") for i in x])
moviesandcredit_df['cast']=moviesandcredit_df['cast'].apply(lambda x:[i.replace(" ","") for i in x])
moviesandcredit_df['crew']=moviesandcredit_df['crew'].apply(lambda x:[i.replace(" ","") for i in x])
```

```python
moviesandcredit_df
```

```python
moviesandcredit_df['tags']=moviesandcredit_df['overview']+moviesandcredit_df['genres']+moviesandcredit_df['keywords']+moviesandcredit_df['cast']+moviesandcredit_df['crew']
```

```python
moviesandcredit_df
```

```python
new_df=moviesandcredit_df[['id','title','tags']].reset_index(drop=True)
new_df
```

```python
new_df['tags']=new_df['tags'].apply(lambda x:" ".join(x))
```

```python
new_df['tags'].head(1).values
```

```python
new_df.head(1)
```

```python
new_df['tags']=new_df['tags'].apply(lambda x:x.lower())
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv=CountVectorizer(max_features=5000,stop_words='english')
```

```python
vectors=cv.fit_transform(new_df['tags']).toarray()
```

```python
vectors
```

```python
cv.get_feature_names_out()
```

```python
import nltk
```

```python
from nltk.stem.porter import PorterStemmer
ps=PorterStemmer()
```

```python
def stem(text):
  y=[]
  for i in text.split():
    y.append(ps.stem(i))
  return " ".join(y)
```

```python
new_df['tags']=new_df['tags'].apply(stem)
new_df['tags'][0]
```

```python
cv.get_feature_names_out()
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity=cosine_similarity(vectors)
```

```python
similarity[0]
```

```python
new_df
```

```python
new_df[new_df['title']=='Newlyweds'].index
```

```python
def recommend(movie):
  movie_index=new_df[new_df['title']==movie].index[0]
  distances=similarity[movie_index]

  movies_similiar=sorted(list(enumerate(distances)),reverse=True,key=lambda x:x[1])[1:5]
  for i in movies_similiar:
    print(new_df.iloc[i[0]].title)
```

```python
recommend('Newlyweds')
```
